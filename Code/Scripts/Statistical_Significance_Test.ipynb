# Statistical Comparison
This script analyzes the statistical significance between raw and quality refined data

```python
# imports
from scipy.stats import shapiro
from scipy.stats import ttest_ind, mannwhitneyu
```

## MDB

```python
raw = [[0.6783, 0.6753, 0.6806, 0.6479, 0.6605], 
    [0.6819, 0.6810, 0.6888, 0.6614, 0.6694], 
    [0.6862, 0.6775, 0.6760, 0.6459, 0.6593], 
    [0.6777, 0.6746, 0.6801, 0.6473, 0.6599], 
    [0.6789, 0.6759, 0.6812, 0.6485, 0.6610],
    [0.7259, 0.7175, 0.7152, 0.6686, 0.6909]]

prep = [[0.7029, 0.7091, 0.7002, 0.7128, 0.7046],
    [0.7033, 0.7114, 0.7107, 0.7178, 0.7082], 
    [0.7061, 0.7143, 0.6952, 0.7110, 0.7081],
    [0.7024, 0.7086, 0.6996, 0.7123, 0.7041],
    [0.7034, 0.7097, 0.7008, 0.7134, 0.7051],
    [0.7500, 0.7558, 0.7387, 0.7539, 0.7505]]

for i in range (0,5):

    print("Shapiro test for raw Dataset:", shapiro(raw[i]))
    print("Shapiro test for pre Dataset:", shapiro(prep[i]))
```

```python
for i in range (0,6):
    stat, p = ttest_ind(raw[i], prep[i], equal_var=False)
    print(f"Welch’s t-test: stat={stat:.3f}, p={p:.4f}")
```

```python
for i in range (0,6):
    stat, p_two_sided = ttest_ind(prep[i], raw[i], equal_var=False)
    # For one-sided p-value (preprocessed > raw)
    if stat > 0:
        p_one_sided = p_two_sided / 2
    else:
        p_one_sided = 1 - (p_two_sided / 2)

    print(f"One-sided p-value: {p_one_sided:.4f}")
```

```python
for i in range (0,6):
    # Mann-Whitney U test
    stat_mw, p_mw = mannwhitneyu(raw[i], prep[i], alternative='two-sided')
    print(f"Mann-Whitney U test p-value: {p_mw:.4f}")
```

```python
for i in range (0,6):
    stat, p_one_sided = mannwhitneyu(prep[i], raw[i], alternative='greater')
    print(f"One-sided p-value: {p_one_sided:.4f}")
```
